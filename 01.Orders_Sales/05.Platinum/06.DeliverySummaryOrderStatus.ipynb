```python
CATALOG = "ORDERS_SALES"
SCHEMA = "GOLD"
TABLE = "SUMMARY_BY_ORDERS_STATUS"

dbutils.widgets.text("order_status")
dbutils.widgets.text("order_status_group")

order_status = dbutils.widgets.get("order_status")
order_status_group = dbutils.widgets.get("order_status_group")

qry = f"""
select 
  order_status,
  order_status_group,
  COUNT(order_delivered_customer_date) AS total_orders,
  to_date(order_purchase_timestamp) AS order_purchase_date,
  to_date(order_approved_at) AS order_approved_date,
  to_date(order_delivered_customer_date) AS order_delivered_customer_date
FROM ORDERS_SALES.SILVER.orders_enriched
GROUP BY order_status, order_status_group, order_purchase_date, order_approved_date, order_delivered_customer_date
"""

result_df = spark.sql(qry)

target_table = f'{CATALOG}.{SCHEMA}.{TABLE}'

result_df.write.format("delta").mode("append").saveAsTable(target_table)
```
