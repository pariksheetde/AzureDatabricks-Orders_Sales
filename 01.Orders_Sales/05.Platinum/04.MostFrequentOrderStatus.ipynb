```python
CATALOG = 'ORDERS_SALES'
SCHEMA = 'GOLD'

dbutils.widgets.text("order_status", "")
dbutils.widgets.text("status_count", "")

order_status = dbutils.widgets.get("order_status").strip().upper()
status_count = dbutils.widgets.get("status_count").strip()

qry = f'''
SELECT
  order_status,
  order_status_group,
  COUNT(*) AS status_count
FROM ORDERS_SALES.SILVER.orders_enriched
WHERE upper(order_status) = '{order_status}'
AND order_status IS NOT NULL
GROUP BY order_status, order_status_group
ORDER BY status_count DESC
'''

result_df = spark.sql(qry)

table_name = f'{CATALOG}.{SCHEMA}.frequent_order_status_{order_status}'

result_df.write.mode('overwrite').saveAsTable(table_name)

print(f'Most Successful Order Status: {order_status}')
print(f'Frequency: {status_count}')

print(f'Table created: {table_name}')
```
