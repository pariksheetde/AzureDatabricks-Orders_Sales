```python
from pyspark.sql.functions import *

dbutils.widgets.text("order_status", "Delivered")
order_status = dbutils.widgets.get("order_status").strip().upper()
CATALOG = 'ORDERS_SALES'
SCHEMA = 'GOLD'

table_name = f'{CATALOG}.{SCHEMA}.STATUS_{order_status}'

qry = f'''
CREATE OR REPLACE TABLE {table_name}
AS
SELECT
  order_status,
  order_status_group,
  COUNT(*) AS total_order_status
FROM ORDERS_SALES.SILVER.orders_enriched
WHERE upper(order_status) = '{order_status}'
GROUP BY order_status, order_status_group
'''

spark.sql(qry)

print(f'Created Table : {table_name}')
```
