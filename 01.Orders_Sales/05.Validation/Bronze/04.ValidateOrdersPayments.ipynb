```sql
-- VALIDATE RECORD COUNT FROM `ORDERS_SALES.BRONZE.ORDERS_PAYMENTS`
SELECT COUNT(*) AS record_cnt,
file_path,
to_date(load_timestamp, 'dd-MM-yyyy') as date
FROM ORDERS_SALES.BRONZE.ORDERS_PAYMENTS
GROUP by file_path, load_timestamp;
```

```sql
-- VALIDATE SAMPLE RECORDS FROM `ORDERS_SALES.BRONZE.CUSTOMERS`
SELECT
*
FROM
ORDERS_SALES.BRONZE.ORDERS_PAYMENTS
LIMIT 3
```

```sql
-- VALIDATE SAMPLE RECORDS FROM `ORDERS_SALES.BRONZE.CUSTOMERS`
SELECT
DISTINCT file_name
FROM
ORDERS_SALES.BRONZE.ORDERS_PAYMENTS
ORDER BY file_name;
```
