#### READ THE CUSTOMERS DATA FROM VOLUME

```python
BASE_DIR = '/Volumes/orders_sales/landing/operational/customers/'
```

```python
cust_schema = """customer_id STRING, 
                 customer_unique_id STRING, 
                 customer_zip_code_prefix INT, 
                 customer_city STRING, 
                 customer_state STRING"""
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

cust_df = (
            spark.read.format('csv')
                 .option('header', True)
                 .schema(cust_schema)
                 .load(f'{BASE_DIR}')
        )
cust_trans_df = cust_df.select("*", "_metadata.file_name", "_metadata.file_path").withColumn('load_timestamp', current_timestamp())
cust_trans_df.printSchema()
```

```python
cust_trans_df.limit(1).display()
```

#### LOAD THE DATA INTO BRONZE LAYER
- `ORDERS_SALES.BRONZE.CUSTOMERS`

```python
cust_trans_df.write.mode('append').format('delta').saveAsTable('ORDERS_SALES.BRONZE.CUSTOMERS')
```
