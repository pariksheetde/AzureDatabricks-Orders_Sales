```sql
-- VALIDATE ORDERS_SALES.AUDIT.PIPELINE FOR BRONZE LAYER
SELECT
*
FROM
ORDERS_SALES.AUDIT.PIPELINE_AUDIT_LOG
WHERE pipeline_name IN ('CUSTOMERS_LOAD', 
                       'GEO_LOCATIONS_LOAD',
                       'ORDER_ITEMS_LOAD',
                       'ORDER_PAYMENTS_LOAD', 
                       'ORDER_REVIEWS_LOAD', 
                       'ORDERS_LOAD', 
                       'PRODUCT_CATEGORY_TRANSLATION_LOAD',
                       'PRODUCTS_LOAD', 
                       'SELLERS_LOAD')
ORDER BY start_time ASC;
```
