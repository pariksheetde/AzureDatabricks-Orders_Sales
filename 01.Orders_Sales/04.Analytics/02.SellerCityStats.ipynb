```sql
CREATE OR REPLACE TABLE ORDERS_SALES.GOLD.SELLER_CITY_STATS
AS
WITH city_counts
AS
(
    SELECT
    seller_city,
    seller_state,
    COUNT(*) AS seller_count
    FROM
    ORDERS_SALES.SILVER.SELLERS_ENRICHED
    GROUP BY seller_city, seller_state
),
ranked AS
(
    SELECT
    seller_city,
    seller_state,
    seller_count,
    DENSE_RANK() OVER (ORDER BY seller_count DESC, seller_city) AS rnk,
    SUM(seller_count) OVER() AS total_all
    FROM city_counts
)
SELECT 
seller_city,
seller_state,
seller_count,
rnk
FROM ranked;
```

```sql
-- VALIDATE ORDERS_SALES.GOLD.SELLER_CITY_STATS
SELECT
*
FROM
ORDERS_SALES.GOLD.SELLER_CITY_STATS
LIMIT 15
```
