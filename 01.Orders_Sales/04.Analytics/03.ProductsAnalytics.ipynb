```sql
CREATE OR REPLACE TABLE ORDERS_SALES.GOLD.PRODUCT_CATEGORY_SUMMARY
WITH product_analytics AS (
    SELECT
        product_category_name,
        ROUND(AVG(product_weight_g),2) AS avg_weight_gm,
        ROUND(AVG(product_length_cm),2) AS avg_length_cm,
        ROUND(AVG(product_height_cm),2) AS avg_height_cm,
        ROUND(AVG(product_width_cm),2) AS avg_width_cm,
        ROUND(AVG(product_volume),2) AS avg_volume,
        ROUND(AVG(product_weight_g) / AVG(product_volume),2) AS avg_density
    FROM
        ORDERS_SALES.SILVER.PRODUCTS_ENRICHED
        WHERE product_category_name IS NOT NULL
        GROUP BY product_category_name
        ORDER BY avg_weight_gm DESC
)
SELECT * FROM product_analytics
```
