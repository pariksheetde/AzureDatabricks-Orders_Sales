- CONTAINER NAME: orderssales
- ACCOUNT NAME: orderssalesdevadls

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS sales_orders_external_location
URL 'abfss://orderssales@orderssalesdevadls.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `orderssales-dev-storage-credential`)
COMMENT 'CREATING EXTERNAL LOCATION FOR ORDERS SALES CONTAINER'
```

#### CREATE SCHEMA
- LANDING
- BRONZE
- SILVER
- GOLD
- AUDIT

```python
%sql
CREATE CATALOG IF NOT EXISTS orders_sales
MANAGED LOCATION 'abfss://orderssales@orderssalesdevadls.dfs.core.windows.net/'
COMMENT 'CREATING CATALOG FOR ORDERS SALES'
```

```python
%sql
USE CATALOG orders_sales;
SELECT current_catalog() as current_catalog;
```

#### CREATE 'LANDING' SCHEMA

```python
%sql
CREATE SCHEMA IF NOT EXISTS orders_sales.landing
COMMENT 'CREATE LANDING SCHEMA'
MANAGED LOCATION 'abfss://orderssales@orderssalesdevadls.dfs.core.windows.net/landing';
```
